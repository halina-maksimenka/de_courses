**Проставьте название соответсвтуещей группы для каждого человека в новой колонке AgeGroup**

**14. Сгруппируйте людей по типу занятости (колонка occupation) и определите количество людей в каждой группе. После чего напишите функциюю фильтрации filter_func, которая будет возвращать только те группы, в которых средний возраст (колонка age) не больше 40 и в которых все работники отрабатывают более 5 часов в неделю (колонка hours-per-week)**

```python
grouped_occupation = df.groupby('occupation')[['age' ,'hours-per-week']].describe().reset_index()
grouped_occupation
```

```python
# создадим функцию, которая возвращает только те группы, которые удовлетворяют условию
def filter_func():
    # список только тех занятостей, у которых средний возраст людей не больше сорока
    groupd_mean_age = grouped_occupation[grouped_occupation['age']['mean'] <= 40]
    list_mean_age = groupd_mean_age.occupation.unique().tolist()
    
    # список только тех занятостей, у которых все работники отрабатывают более 5 часов в неделю
    # (значит минимальное значение для группы людей должно быть выше 5 часов)
    groupd_min_hours = grouped_occupation[grouped_occupation['hours-per-week']['min'] > 5]
    list_min_hours = groupd_min_hours.occupation.unique().tolist()
    
    # Находим совпадающие значения в двух списках
    common_occupations = [item for item in list_mean_age if item in list_min_hours]
    
    return common_occupations
```

```python
list_occupations = filter_func()
list_occupations
```

```python
result_df = df[df['occupation'].isin(list_occupations)]
result_df
```
